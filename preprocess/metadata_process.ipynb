```python
import pandas as pd 
import numpy as np
```

```python
file1 = "/mnt/data/tuongnv/Forecast/data/metadata.csv"
df_meta = pd.read_csv(file1)
df_meta.columns
```

```python
# chilledwater_cleaned
df_features = df_meta[['electricity', 'hotwater', 'chilledwater', 'steam',
       'water', 'irrigation', 'solar', 'gas', 'industry', 'subindustry',
       'heatingtype', 'numberoffloors']]
count_values = df_features.count()
print(count_values)
```

```python
df_features['numberoffloors'] = np.where(df_features['numberoffloors'].notna(), 'Yes', df_features['numberoffloors'])
df_features['numberoffloors']
```

```python

import pandas as pd
import itertools

cols = [
    "electricity", "hotwater", "chilledwater", "steam",
    "water", "irrigation", "solar", "gas",
    "industry", "subindustry", "heatingtype", "numberoffloors",
]
# cols = [
#     "electricity", "hotwater", "chilledwater", "numberoffloors"
# ]

results = []

# duyệt size tổ hợp từ 2 trở lên
for r in range(2, len(cols) + 1):
    for combo in itertools.combinations(cols, r):
        
        # mask: tất cả cột trong combo == 'Yes'
        mask = df_features[list(combo)].eq("Yes").all(axis=1)
        count = mask.sum()

        if count > 0:
            results.append({
                "combination": combo,
                "num_rows": int(count)
            })

df_result = pd.DataFrame(results)
df_result = df_result.sort_values(
    by="num_rows",
    ascending=False
).reset_index(drop=True)

# df_result[:20]
df_result.to_csv("/mnt/data/tuongnv/Forecast/preprocess/merge_feature.csv")
```
